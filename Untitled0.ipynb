```python
#-----------------------------------------------
# MESA DE PARTES DIGITAL - BOTICA LIMA S.A.C
#-----------------------------------------------

def mostrar_menu():
    """Muestra el menú principal en pantalla."""
    print('-------------------------------------------')
    print(' mesa de partes digital_BOTICA LIMA S.A.C ')
    print('-------------------------------------------')
    print(' 1. REGISTRAR EXPEDIENTE')
    print(' 2. CONSULTAR EXPEDIENTE POR CODIGO')
    print(' 3. LISTAR Y ORDENAR EXPEDIENTE')
    print(' 4. SALIR')

def registrar_expediente(codigos, dnis, nombres, asuntos):
    """Función para registrar un nuevo expediente validando el código."""
    print('\n=== REGISTRAR NUEVO EXPEDIENTE ===')
    aux = input('Ingrese codigo (Formato: EXP-XXXX): ')

    if len(aux) != 8:
        print('Error: El codigo debe tener 8 caracteres.')
    else:
        codigos.append(aux)
        dnis.append(input('Ingrese DNI del ciudadano: '))
        nombres.append(input('Ingrese nombre completo: '))
        asuntos.append(input('Ingrese el asunto: '))
        print('¡Expediente registrado con exito en memoria!')

def consultar_expediente(codigos, dnis, nombres, asuntos):
    """Función para buscar un expediente mediante su código."""
    print('\n=== CONSULTAR EXPEDIENTE ===')
    aux = input('Ingrese el codigo a buscar: ')
    encontrado = False

    for i in range(len(codigos)):
        if codigos[i] == aux:
            print(f'Codigo: {codigos[i]}')
            print(f'DNI:    {dnis[i]}')
            print(f'Nombre: {nombres[i]}')
            print(f'Asunto: {asuntos[i]}')
            encontrado = True
            break

    if not encontrado:
        print('No se encontro ningun expediente con ese codigo.')

def ordenar_y_listar_expedientes(codigos, dnis, nombres, asuntos):
    print('\n=== LISTADO ORDENADO POR METODO BURBUJA ===')
    cantidad_expedientes = len(codigos)

    if cantidad_expedientes == 0:
        print('La mesa de partes esta vacia.')
    else:
        if cantidad_expedientes > 1:
            for i in range(cantidad_expedientes - 1):
                for j in range(cantidad_expedientes - 1 - i):
                    if codigos[j] > codigos[j+1]:

                        codigos[j], codigos[j+1] = codigos[j+1], codigos[j]

                        dnis[j], dnis[j+1] = dnis[j+1], dnis[j]

                        nombres[j], nombres[j+1] = nombres[j+1], nombres[j]

                        asuntos[j], asuntos[j+1] = asuntos[j+1], asuntos[j]

        print("CODIGO: |  DNI:  |   NOMBRE:   |    ASUNTO:  ")
        for i in range(cantidad_expedientes):
            print(f'{codigos[i]} | {dnis[i]} | {nombres[i]} | {asuntos[i]}')

def main():
    codigos = []
    dnis = []
    nombres = []
    asuntos = []
    opcion = 0

    while opcion != 4:
        mostrar_menu()
        opcion_input = input(' seleccione una opcion (1-4) :')

        if opcion_input.isdigit():
            opcion = int(opcion_input)

            if opcion == 1:
                registrar_expediente(codigos, dnis, nombres, asuntos)
                input('presiona enter para continuar...')

            elif opcion == 2:
                consultar_expediente(codigos, dnis, nombres, asuntos)
                input('presiona enter para continuar...')

            elif opcion == 3:
                ordenar_y_listar_expedientes(codigos, dnis, nombres, asuntos)
                input('presiona enter para continuar...')

            elif opcion == 4:
                print('Saliendo del sistema...')

            else:
                print('Opcion incorrecta. Debe ser un numero entre 1 y 4.')
                input('\npresiona enter para continuar...')
        else:
            print('Opcion incorrecta. Por favor, ingrese un numero valido.')
            input('presiona enter para continuar...')

if __name__ == "__main__":
    main()


```
